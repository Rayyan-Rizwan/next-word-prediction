```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("qoute_dataset.csv")
```

```python
# Check dataset structure
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns)
print("\nColumn data types:")
print(df.dtypes)
print("\nFirst few rows to identify text column:")
print(df.head())
```

```python
# Select the quote column
quotes = df['quote']

# Display some basic info
print("Quote column shape:", quotes.shape)
print("\nFirst 5 quotes:")
print(quotes.head())
print("\nSample quotes:")
print(quotes.sample(3).values)
```

```python
import re
import string

# Step 2.1: Remove NaN values
quotes_clean = quotes.dropna()
print("After removing NaN values:", quotes_clean.shape)

# Step 2.2: Convert to lowercase
quotes_clean = quotes_clean.str.lower()

# Step 2.3: Remove punctuation and extra spaces
def clean_text(text):
    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # Remove extra spaces
    text = ' '.join(text.split())
    return text

quotes_clean = quotes_clean.apply(clean_text)

# Step 2.4: Convert to Python list
quotes_list = quotes_clean.tolist()

# Display results
print("\nFirst 5 cleaned quotes:")
for i, quote in enumerate(quotes_list[:5], 1):
    print(f"{i}. {quote}")

print("\nTotal quotes after cleaning:", len(quotes_list))
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

# Step 3.1: Create and fit the tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(quotes_list)

# Step 3.2: Get the vocabulary (word to index dictionary)
word_index = tokenizer.word_index
print("Vocabulary size (unique words):", len(word_index))
print("\nFirst 20 words in vocabulary:")
for i, (word, idx) in enumerate(list(word_index.items())[:20], 1):
    print(f"{i}. {word}: {idx}")

# Step 3.3: Convert text to sequences of numerical IDs
sequences = tokenizer.texts_to_sequences(quotes_list)
print("\nFirst 5 tokenized sequences:")
for i, seq in enumerate(sequences[:5], 1):
    print(f"{i}. {seq}")

# Step 3.4: Check sequence lengths
seq_lengths = [len(seq) for seq in sequences]
print("\nSequence length statistics:")
print(f"Min length: {min(seq_lengths)}")
print(f"Max length: {max(seq_lengths)}")
print(f"Average length: {np.mean(seq_lengths):.2f}")
```

```python
# Step 4: Generate N-gram sequences
ngram_sequences = []

for quote in quotes_list:
    # Split quote into words
    words = quote.split()

    # Skip if quote has less than 2 words
    if len(words) < 2:
        continue

    # Create n-gram sequences (progressive sequences)
    for i in range(1, len(words)):
        # Get sequence from start to current position
        ngram = words[:i+1]
        ngram_sequences.append(ngram)

# Display results
print("Total N-gram sequences generated:", len(ngram_sequences))
print("\nFirst 10 N-gram sequences:")
for i, ngram in enumerate(ngram_sequences[:10], 1):
    print(f"{i}. {ngram}")

print("\nLast 5 N-gram sequences:")
for i, ngram in enumerate(ngram_sequences[-5:], len(ngram_sequences)-4):
    print(f"{i}. {ngram}")

# Check sequence lengths
ngram_lengths = [len(ngram) for ngram in ngram_sequences]
print(f"\nN-gram length statistics:")
print(f"Min length: {min(ngram_lengths)}")
print(f"Max length: {max(ngram_lengths)}")
print(f"Average length: {np.mean(ngram_lengths):.2f}")
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Step 5.1: Convert n-gram sequences to numerical sequences
input_sequences = []
for ngram in ngram_sequences:
    # Convert words to numerical IDs using tokenizer
    sequence = tokenizer.texts_to_sequences([' '.join(ngram)])[0]
    input_sequences.append(sequence)

# Step 5.2: Pad sequences to make them equal length
max_sequence_len = max([len(seq) for seq in input_sequences])
padded_sequences = pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')

print("Padded sequences shape:", padded_sequences.shape)
print("\nFirst 5 padded sequences:")
print(padded_sequences[:5])

# Step 5.3: Split into X (input) and y (target)
X = padded_sequences[:, :-1]  # All words except the last one
y = padded_sequences[:, -1]   # Only the last word (target to predict)

print("\nInput features (X) shape:", X.shape)
print("Target labels (y) shape:", y.shape)
print("\nFirst 5 input sequences (X):")
print(X[:5])
print("\nFirst 5 target labels (y):")
print(y[:5])

# Step 5.4: Get vocabulary size for model
vocab_size = len(tokenizer.word_index) + 1  # +1 for padding token
print("\nVocabulary size:", vocab_size)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Input
from tensorflow.keras.optimizers import Adam

# Step 6.1: Define model architecture with proper Input layer
model = Sequential()

# Input Layer (replaces deprecated input_length parameter)
model.add(Input(shape=(X.shape[1],)))

# Embedding Layer: Convert words to dense vectors
model.add(Embedding(input_dim=vocab_size, output_dim=100))

# First LSTM Layer with Dropout
model.add(LSTM(150, return_sequences=True))
model.add(Dropout(0.2))

# Second LSTM Layer with Dropout
model.add(LSTM(100))
model.add(Dropout(0.2))

# Dense Layer with Softmax activation for prediction
model.add(Dense(vocab_size, activation='softmax'))

# Step 6.2: Compile the model
model.compile(optimizer=Adam(learning_rate=0.001),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Step 6.3: Display model summary (now shows actual parameters)
print("Model Architecture:")
print("=" * 50)
model.summary()

# Step 6.4: Display layer details (fixed - using get_config instead)
print("\nModel Layers:")
for i, layer in enumerate(model.layers, 1):
    print(f"{i}. {layer.name}: {layer.__class__.__name__}")

# Verify model is built
print("\nModel built successfully:", model.built)
print("Total parameters:", model.count_params())
```

```python
# Step 7.1: Model is already compiled in Step 6
# Verify compilation
print("Model compilation status:")
print("Optimizer:", model.optimizer)
print("Loss function:", model.loss)
print("Metrics:", model.metrics)

# Step 7.2: Train the model
# Define training parameters
epochs = 100
batch_size = 64

print("\n" + "="*50)
print("Starting Model Training")
print("="*50)
print(f"Training samples: {X.shape[0]}")
print(f"Input shape: {X.shape}")
print(f"Target shape: {y.shape}")
print(f"Epochs: {epochs}")
print(f"Batch size: {batch_size}")
print("="*50 + "\n")

# Step 7.3: Fit the model
history = model.fit(X, y,
                    epochs=epochs,
                    batch_size=batch_size,
                    verbose=1)

# Step 7.4: Display training results
print("\n" + "="*50)
print("Training Completed!")
print("="*50)
print(f"Final Training Loss: {history.history['loss'][-1]:.4f}")
print(f"Final Training Accuracy: {history.history['accuracy'][-1]:.4f}")
```

```python
import numpy as np

# Step 8.1: Function to predict next word
def predict_next_word(model, tokenizer, text, max_sequence_len):
    """
    Predict the next word given a text prompt.

    Parameters:
    - model: Trained Keras model
    - tokenizer: Fitted Keras Tokenizer
    - text: Input text prompt (string)
    - max_sequence_len: Maximum sequence length used during training

    Returns:
    - Predicted next word (string)
    """
    # Convert text to lowercase
    text = text.lower()

    # Tokenize the input text
    token_list = tokenizer.texts_to_sequences([text])[0]

    # Pad the sequence to match training length
    token_list = pad_sequences([token_list], maxlen=max_sequence_len, padding='pre')[0]

    # Reshape for model input (1, sequence_length)
    token_list = token_list.reshape(1, -1)

    # Predict probabilities for all words in vocabulary
    predictions = model.predict(token_list, verbose=0)[0]

    # Get the index of the word with highest probability
    predicted_index = np.argmax(predictions)

    # Convert index back to word
    for word, index in tokenizer.word_index.items():
        if index == predicted_index:
            return word

    return None

# Step 8.2: Function to generate complete sentence
def generate_sentence(model, tokenizer, prompt, max_sequence_len, num_words=1):
    """
    Generate a complete sentence by predicting multiple words.

    Parameters:
    - model: Trained Keras model
    - tokenizer: Fitted Keras Tokenizer
    - prompt: Initial text prompt from user (string)
    - max_sequence_len: Maximum sequence length used during training
    - num_words: Number of words to generate (default: 1)

    Returns:
    - Generated sentence (string)
    """
    # Start with the user's prompt
    result_text = prompt.lower()

    # Generate specified number of words
    for _ in range(num_words):
        # Predict next word
        next_word = predict_next_word(model, tokenizer, result_text, max_sequence_len)

        # If no word predicted, break
        if next_word is None:
            break

        # Append predicted word to result
        result_text = result_text + " " + next_word

    return result_text

# Step 8.3: Test the prediction function
print("="*50)
print("Testing Prediction Function")
print("="*50)

# Example prompts
test_prompts = [
    "life is",
    "the world",
    "love is"
]

for prompt in test_prompts:
    predicted_word = predict_next_word(model, tokenizer, prompt, X.shape[1])
    print(f"\nPrompt: '{prompt}'")
    print(f"Predicted next word: '{predicted_word}'")

# Step 8.4: Test sentence generation
print("\n" + "="*50)
print("Testing Sentence Generation")
print("="*50)

for prompt in test_prompts:
    generated = generate_sentence(model, tokenizer, prompt, X.shape[1], num_words=3)
    print(f"\nPrompt: '{prompt}'")
    print(f"Generated: '{generated}'")
```
